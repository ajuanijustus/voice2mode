```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# SVM

## HuBERT

```python
hubert_5fold_df = pd.read_csv('results/hubert_5fold_df_mean_embd.csv')[['Layer', 'Fold', 'Fold_Accuracy']]
hubert_5fold_df[hubert_5fold_df['Fold_Accuracy']==hubert_5fold_df['Fold_Accuracy'].max()]
```

```python
max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].agg(['mean', 'std']).max()
index_of_max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean().idxmax()

print("Maximum mean accuracy:", max_mean_accuracy['mean'])
print("Standard deviation of maximum mean accuracy:", max_mean_accuracy['std'])
print("Index of maximum mean accuracy (Layer):", index_of_max_mean_accuracy)
```

```python
# Calculate mean and standard deviation
mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()
std_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].std()

# Convert to DataFrame
data = pd.DataFrame({'Layer': mean_accuracy.index, 'Mean Accuracy': mean_accuracy, 'Standard Deviation': std_accuracy})

# Plot using Seaborn
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.lineplot(data=data, x='Layer', y='Mean Accuracy', marker='o', label='Mean Accuracy')
sns.scatterplot(data=data, x='Layer', y='Mean Accuracy', hue='Layer', legend=False)

# Add error bars
plt.errorbar(data['Layer'], data['Mean Accuracy'], yerr=data['Standard Deviation'], fmt='o', label='Standard Deviation')

# Add labels and title
plt.xlabel('Layer')
plt.ylabel('Accuracy')
plt.title('HuBERT: Mean Accuracy with Standard Deviation')

# Show plot
plt.legend()
plt.show()
```
