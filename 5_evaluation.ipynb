```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
import seaborn as sns
```

# SVM

## HuBERT

```python
hubert_5fold_df = pd.read_csv('results/hubert_5fold_df_mean_embd.csv')[['Layer', 'Fold', 'Fold_Accuracy']]
hubert_5fold_df[hubert_5fold_df['Fold_Accuracy']==hubert_5fold_df['Fold_Accuracy'].max()]
```

```python
max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].agg(['mean', 'std']).max()
index_of_max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean().idxmax()

print("Maximum mean accuracy:", max_mean_accuracy['mean'])
print("Standard deviation of maximum mean accuracy:", max_mean_accuracy['std'])
print("Index of maximum mean accuracy (Layer):", index_of_max_mean_accuracy)
```

```python
# Calculate mean and standard deviation
mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()
std_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].std()

# Convert to DataFrame
data = pd.DataFrame({'Layer': mean_accuracy.index, 'Mean Accuracy': mean_accuracy, 'Standard Deviation': std_accuracy})

# Plot using Seaborn
sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
sns.lineplot(data=data, x='Layer', y='Mean Accuracy', marker='o', label='Mean Accuracy')
sns.scatterplot(data=data, x='Layer', y='Mean Accuracy', hue='Layer', legend=False)

# Add error bars
plt.errorbar(data['Layer'], data['Mean Accuracy'], yerr=data['Standard Deviation'], fmt='o', label='Standard Deviation')

# Add labels and title
plt.xlabel('Layer')
plt.ylabel('Accuracy')
plt.title('HuBERT: Mean Accuracy with Standard Deviation')

# Show plot
plt.legend()
plt.show()
```

```python
# Load your dataframes (hubert_5fold_df, wav2vec2base_5fold_df, wav2vec2large_5fold_df)

# Calculate mean and standard deviation for each model
hubert_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()
hubert_std_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].std()

wav2vec2base_mean_accuracy = wav2vec2base_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()
wav2vec2base_std_accuracy = wav2vec2base_5fold_df.groupby('Layer')['Fold_Accuracy'].std()

wav2vec2large_mean_accuracy = wav2vec2large_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()
wav2vec2large_std_accuracy = wav2vec2large_5fold_df.groupby('Layer')['Fold_Accuracy'].std()

# Combine dataframes
data = pd.concat([
    pd.DataFrame({'Layer': hubert_mean_accuracy.index, 'Mean Accuracy': hubert_mean_accuracy, 'Standard Deviation': hubert_std_accuracy, 'Model': 'HuBERT'}),
    pd.DataFrame({'Layer': wav2vec2base_mean_accuracy.index, 'Mean Accuracy': wav2vec2base_mean_accuracy, 'Standard Deviation': wav2vec2base_std_accuracy, 'Model': 'wav2vec2-BASE'}),
    pd.DataFrame({'Layer': wav2vec2large_mean_accuracy.index, 'Mean Accuracy': wav2vec2large_mean_accuracy, 'Standard Deviation': wav2vec2large_std_accuracy, 'Model': 'wav2vec2-LARGE'})
])

# Plot using Seaborn
plt.figure(figsize=(10, 6))
sns.lineplot(data=data, x='Layer', y='Mean Accuracy', hue='Model', marker='o', err_style='bars', errorbar='sd', palette='colorblind')

# Convert y-axis to percentages
plt.gca().yaxis.set_major_formatter(ticker.FuncFormatter(lambda y, _: '{:.0%}'.format(y)))
# Show every value on the x-axis
plt.xticks(data['Layer'])

# Add error bars
for i, model in enumerate(['HuBERT', 'wav2vec2-BASE', 'wav2vec2-LARGE']):
    model_data = data[data['Model'] == model]
    plt.errorbar(model_data['Layer'], model_data['Mean Accuracy'], yerr=model_data['Standard Deviation'], fmt='o', color=sns.color_palette('colorblind')[i])

# Add labels and title
plt.xlabel('Layer')
plt.ylabel('Accuracy')
plt.title('Layer-wise Classification Accuracies for Foundation Models using SVM')

# Show plot
plt.legend()
plt.grid(True, alpha=0.2)

# Save plot as SVG
plt.savefig('results/cm_svm.svg', format='svg')

plt.show()
```
