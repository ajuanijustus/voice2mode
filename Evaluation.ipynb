```python
import pandas as pd
import matplotlib.pyplot as plt
```

## HuBERT

```python
hubert_5fold_df = pd.read_csv('hubert_5fold_df.csv')[['Layer', 'Fold', 'Fold_Accuracy']]
hubert_5fold_df[hubert_5fold_df['Fold_Accuracy']==hubert_5fold_df['Fold_Accuracy'].max()]
```

```python
max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean().max()
index_of_max_mean_accuracy = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean().idxmax()

print("Maximum mean accuracy:", max_mean_accuracy)
print("Index of maximum mean accuracy (Layer):", index_of_max_mean_accuracy)
```

```python
# Grouping by 'Layer' and calculating the mean of 'Fold_Accuracy'
mean_accuracy_by_layer = hubert_5fold_df.groupby('Layer')['Fold_Accuracy'].mean()

# Plotting the mean fold accuracy
plt.figure(figsize=(8, 6))
mean_accuracy_by_layer.plot(kind='bar', color='skyblue')
plt.title('Mean Fold Accuracy by Layer')
plt.xlabel('Layer')
plt.ylabel('Mean Fold Accuracy')
plt.xticks(rotation=45)  # Rotate x-axis labels for better readability if necessary
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
